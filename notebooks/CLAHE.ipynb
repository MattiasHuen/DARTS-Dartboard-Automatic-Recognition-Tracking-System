# **CLAHE section**

```python
from pathlib import Path
import cv2 
import numpy as np
import matplotlib.pyplot as plt
```

```python

def load_image(url: str = "..\\notebooks\\test_images\\lenna_(test_image).png"):
    """
    This function is simply for loading an image
    """
    image_path = Path(url).resolve()
    
    print(f"Looking for video at: {image_path}")
    
    if not image_path.is_file():
        raise FileNotFoundError(f"Video not found: {image_path}")

    image = cv2.imread(str(image_path))
    return cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

img = load_image()
```

## **Example of how CLAHE affects the contrast based on clip limit and grid size**

```python
import cv2
import matplotlib.pyplot as plt

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

clip_limits = [1, 2, 3, 5, 10]
grid_sizes = [(3,3), (5,5), (9,9), (15,15), (21,21)]

fig, axs = plt.subplots(5, 5, figsize=(10,10), layout="constrained")

for row, i in enumerate(clip_limits):
    for col, j in enumerate(grid_sizes):

        clahe = cv2.createCLAHE(clipLimit=i, tileGridSize=j)
        frame_clahe = clahe.apply(img_gray)

        im = axs[row, col].imshow(
            frame_clahe,
            cmap="gray",
            vmin=0,
            vmax=255
        )

        axs[row, col].set_title(f"Clip: {i}, Grid: {j}", fontsize=10)
        axs[row, col].axis("off")

cbar = fig.colorbar(
    im,
    ax=axs,
    orientation="horizontal",
    shrink=0.966,
    pad=0.00,
    ticks=[0, 50, 100, 150, 200, 255]
)

cbar.set_label("Pixel intensity")

plt.show()
```

## **Example of how the clip limit affects the pixel distribution inside each grid**

```python
from matplotlib.patches import Rectangle

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

clip_limit = 2
grid_size = (5, 5)

clahe = cv2.createCLAHE(
    clipLimit=clip_limit,
    tileGridSize=grid_size
)
img_clahe = clahe.apply(img_gray)

# Calculate grid boundaries
h, w = img_gray.shape
nx, ny = grid_size

x_edges = np.linspace(0, w, nx + 1, dtype=int)
y_edges = np.linspace(0, h, ny + 1, dtype=int)

# Select the center grid area
col = nx // 2
row = ny // 2

x0, x1 = x_edges[col], x_edges[col + 1]
y0, y1 = y_edges[row], y_edges[row + 1]

# Extract the same area before and after
tile_before = img_gray[y0:y1, x0:x1]
tile_after = img_clahe[y0:y1, x0:x1]

# Calculate histograms
hist_before = cv2.calcHist([tile_before], [0], None, [256], [0, 256])
hist_after = cv2.calcHist([tile_after], [0], None, [256], [0, 256])

fig, axs = plt.subplots(
    2, 2,
    figsize=(13, 9),
    gridspec_kw={"height_ratios": [3, 1.5]},
    layout="constrained"
)

for ax, image, title in zip(
    axs[0],
    [img_gray, img_clahe],
    ["Original image", "CLAHE image"]
):
    ax.imshow(image, cmap="gray", vmin=0, vmax=255)

    for x in x_edges[1:-1]:
        ax.axvline(x - 0.5, color="yellow", linewidth=0.8)

    for y in y_edges[1:-1]:
        ax.axhline(y - 0.5, color="yellow", linewidth=0.8)

    rect = Rectangle(
        (x0 - 0.5, y0 - 0.5),
        x1 - x0,
        y1 - y0,
        edgecolor="red",
        facecolor="none",
        linewidth=5
    )
    ax.add_patch(rect)

    ax.set_title(title, fontsize = 18)
    ax.axis("off")

axs[1, 0].plot(hist_before, color="blue")
axs[1, 0].set_title("Histogram of Pixel Distribution before CLAHE", fontsize = 16)

axs[1, 1].plot(hist_after, color="red")
axs[1, 1].set_title("Histogram of Pixel Distribution after CLAHE", fontsize = 16)

for ax in axs[1]:
    ax.set_xlim(0, 255)
    ax.set_xlabel("Pixel intensity", fontsize = 12)
    ax.set_ylabel("Pixel count", fontsize = 12)
    ax.grid(alpha=0.3)

max_count = max(hist_before.max(), hist_after.max())
for ax in axs[1]:
    ax.set_ylim(0, max_count * 1.05)

plt.suptitle(
    f"CLAHE: Clip Limit = {clip_limit}, Grid Size = {grid_size}",
    fontsize=15
)

plt.show()
```

# **Automating grid size selection and clahe clip limit**

we calculate the difference between the 20th and 80th percentile to estimate the image's intensity range, which means images with greater intensity variation receive a higher CLAHE clip limit.

```python
def choose_clahe_grid(frame, tile_size=64):
    """
    This function automatically chooses a grid. We use a tile size of 64, beacuse
    we aim for each CLAHE tile to cover approximately 64 x 64 pixels.

    The function also limits the grid size to be between 4 and 16 in each direction.
    """
    h, w = frame.shape[:2]

    grid_x = int(np.clip(round(w / tile_size), 4, 16))
    grid_y = int(np.clip(round(h / tile_size), 4, 16))

    return (grid_x, grid_y)

def choose_clahe_limit(gray):
    """
    Estimate image contrast using the 20th and 80th
    percentiles and adapt the CLAHE clip limit.

    Lower contrast -> higher clip limit
    Higher contrast -> lower clip limit
    """

    p20, p80 = np.percentile(gray, [20, 80])
    contrast = p80 - p20

    clip_limit = np.interp(
        contrast,
        [0, 150],
        [3.0, 1.5]
    )

    return float(clip_limit)

```

```python
from matplotlib.patches import Rectangle

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

clip_limit = choose_clahe_limit(img_gray)
grid_size = choose_clahe_grid(img_gray)

print(clip_limit)
clahe = cv2.createCLAHE(
    clipLimit=clip_limit,
    tileGridSize=grid_size
)
img_clahe = clahe.apply(img_gray)

# Calculate grid boundaries
h, w = img_gray.shape
nx, ny = grid_size

x_edges = np.linspace(0, w, nx + 1, dtype=int)
y_edges = np.linspace(0, h, ny + 1, dtype=int)

# Select the center grid area
col = nx // 2
row = ny // 2

x0, x1 = x_edges[col], x_edges[col + 1]
y0, y1 = y_edges[row], y_edges[row + 1]

# Extract the same area before and after
tile_before = img_gray[y0:y1, x0:x1]
tile_after = img_clahe[y0:y1, x0:x1]

# Calculate histograms
hist_before = cv2.calcHist([tile_before], [0], None, [256], [0, 256])
hist_after = cv2.calcHist([tile_after], [0], None, [256], [0, 256])

fig, axs = plt.subplots(
    2, 2,
    figsize=(13, 9),
    gridspec_kw={"height_ratios": [3, 1.5]},
    layout="constrained"
)

for ax, image, title in zip(
    axs[0],
    [img_gray, img_clahe],
    ["Original image", "CLAHE image"]
):
    ax.imshow(image, cmap="gray", vmin=0, vmax=255)

    for x in x_edges[1:-1]:
        ax.axvline(x - 0.5, color="yellow", linewidth=0.8)

    for y in y_edges[1:-1]:
        ax.axhline(y - 0.5, color="yellow", linewidth=0.8)

    rect = Rectangle(
        (x0 - 0.5, y0 - 0.5),
        x1 - x0,
        y1 - y0,
        edgecolor="red",
        facecolor="none",
        linewidth=5
    )
    ax.add_patch(rect)

    ax.set_title(title, fontsize = 18)
    ax.axis("off")

axs[1, 0].plot(hist_before, color="blue")
axs[1, 0].set_title("Histogram of Pixel Distribution before CLAHE", fontsize = 16)

axs[1, 1].plot(hist_after, color="red")
axs[1, 1].set_title("Histogram of Pixel Distribution after CLAHE", fontsize = 16)

for ax in axs[1]:
    ax.set_xlim(0, 255)
    ax.set_xlabel("Pixel intensity", fontsize = 12)
    ax.set_ylabel("Pixel count", fontsize = 12)
    ax.grid(alpha=0.3)

max_count = max(hist_before.max(), hist_after.max())
for ax in axs[1]:
    ax.set_ylim(0, max_count * 1.05)

plt.suptitle(
    f"CLAHE: Clip Limit = {round(clip_limit,2)}, Grid Size = {grid_size}",
    fontsize=15
)

plt.show()
```

# **Test on videos**

```python
import cv2
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

def get_training_video(url: str = r"..\data\videos\black_background_noHuman.mp4"):
    """
    Loads a training video from a file location.

    Args:
        url: File location

    Returns:
        cv2.VideoCapture object
    """

    video_path = Path(url).resolve()

    print(f"Looking for video at: {video_path}")

    if not video_path.is_file():
        raise FileNotFoundError(f"Video not found: {video_path}")

    video = cv2.VideoCapture(str(video_path))

    if not video.isOpened():
        raise RuntimeError(f"Could not open video: {video_path}")

    fps = video.get(cv2.CAP_PROP_FPS)
    total_frames = video.get(cv2.CAP_PROP_FRAME_COUNT)

    print(f"FPS: {fps}")
    print(f"Total frames: {total_frames}")
    print(f"Total video length: {round(total_frames/fps,2)} seconds")

    return video
```

```python
import cv2
import numpy as np

video_path1 = r"..\data\videos\black_background_yesHuman.mp4"
video_path2 = r"..\data\videos\white_background_yesHuman.mp4"
video_path3 = r"..\data\videos\wood_background_yesHuman.mp4"

videos = [
    get_training_video(video_path1),
    get_training_video(video_path2),
    get_training_video(video_path3)
]


try:
    while True:
        frames = []
        for i, video in enumerate(videos):
            ret, frame = video.read()

            if not ret:
                break

            img_gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
            # Perform CLAHE
            clip_limit = choose_clahe_limit(img_gray)
            grid_size = choose_clahe_grid(img_gray)

            clahe = cv2.createCLAHE(
                clipLimit=clip_limit,
                tileGridSize=grid_size
            )
            frame = clahe.apply(img_gray)

            processed = cv2.resize(frame,
                                   (480, 270))
            frames.append(processed)

        if len(frames) != len(videos):
            break

        for i, frame in enumerate(frames):
            name = f"Video {i+1}"
            cv2.imshow(name, frame)
            cv2.moveWindow(name, i * 500, 100)

        if cv2.waitKey(33) & 0xFF == ord('q'):
            break

finally:
    for video in videos:
        video.release()

    cv2.destroyAllWindows()
```
