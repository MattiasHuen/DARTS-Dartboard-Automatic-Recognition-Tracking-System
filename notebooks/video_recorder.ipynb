# **Video Test Recorder**

This program is for recording a 15 second video to use for testing code

```python

import cv2
import time

# Open laptop camera
cap = cv2.VideoCapture(1)

if not cap.isOpened():
    print("Could not open camera")
    exit()

# Camera settings
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

if fps <= 0 or fps > 120:
    fps = 30.0

# Save video as MP4
fourcc = cv2.VideoWriter_fourcc(*"mp4v")
out = cv2.VideoWriter(
    "darts_video.mp4",
    fourcc,
    fps,
    (width, height)
)

if not out.isOpened():
    cap.release()
    raise RuntimeError("Could not create video file")

# Start recording
start_time = time.monotonic()

while time.monotonic() - start_time < 15:
    ret, frame = cap.read()

    if not ret:
        break

    # Save frame to video
    out.write(frame)

    # Show camera
    cv2.imshow("Darts Camera", frame)

    # Press q to stop early
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

# Release camera and video
cap.release()
out.release()
cv2.destroyAllWindows()

print("Video saved as darts_video.mp4")
```

```python

import cv2
import time
from IPython.display import display, clear_output, Image

# Load video
cap = cv2.VideoCapture("notebooks//white_background_noHuman.mp4")

if not cap.isOpened():
    raise RuntimeError("Could not open video")

fps = cap.get(cv2.CAP_PROP_FPS)
if fps <= 0:
    fps = 30

try:
    while True:
        ret, frame = cap.read()

        if not ret:
            break

        # Convert frame to JPEG for Jupyter
        success, buffer = cv2.imencode(".jpg", frame)

        if not success:
            break

        # Display frame inside notebook
        clear_output(wait=True)
        display(Image(data=buffer.tobytes()))

        time.sleep(1 / fps)

finally:
    cap.release()

print("Video finished")
```
